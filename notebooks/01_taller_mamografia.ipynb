Una mamografía es una imagen de rayos X de la mama. Una **masa** ocupa una región de la imagen; las **calcificaciones** suelen verse como puntos pequeños y claros. Ninguno de esos hallazgos, por sí solo, confirma un diagnóstico.

El recuadro rojo rodea la zona anotada por CBIS-DDSM. Sus cuatro coordenadas están guardadas en `casos.json` y se calcularon a partir de la máscara de referencia. **No lo ha generado una IA.**

```python
from pathlib import Path
import io
import json
import zipfile

import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image

# Busca los datos en este proyecto o en Colab.
carpeta = next(
    (p for p in [Path("."), Path(".."), Path("/content")]
     if (p / "data/taller/casos.json").exists()),
    None,
)

if carpeta is None:
    from google.colab import files
    print("Sube taller_colab.zip")
    archivos = files.upload()
    nombre = next((n for n in archivos if n.endswith(".zip")), None)
    if nombre is None:
        raise ValueError("Selecciona taller_colab.zip para continuar.")
    with zipfile.ZipFile(io.BytesIO(archivos[nombre])) as paquete:
        paquete.extractall("/content")
    carpeta = Path("/content")

casos = json.loads((carpeta / "data/taller/casos.json").read_text(encoding="utf-8"))
print("Listo: tres casos para practicar y ocho en la galería final.")
```

## Una función para mostrar cada caso

Esta celda abre el PNG preparado a partir del DICOM original. Si le das una posición, dibuja tu cuadrado celeste. Si pides la referencia, dibuja un recuadro rojo alrededor de la máscara anotada. Puedes ejecutarla sin cambiar nada.

```python
def mostrar(caso_numero, horizontal=None, vertical=None, referencia=False):
    caso = casos[caso_numero - 1]
    imagen = Image.open(carpeta / "data/taller" / caso["imagen"]).convert("L")
    ancho, alto = imagen.size

    fig, ax = plt.subplots(figsize=(6, 9))
    ax.imshow(imagen, cmap="gray", vmin=0, vmax=255)

    if referencia:
        izquierda, arriba, derecha, abajo = caso["recuadro_referencia"]
        ax.add_patch(Rectangle((izquierda, arriba), derecha - izquierda,
                               abajo - arriba, fill=False,
                               edgecolor="red", linewidth=3))

    if horizontal is not None and vertical is not None:
        x = horizontal / 100 * (ancho - 1)
        y = vertical / 100 * (alto - 1)
        ax.scatter(x, y, s=220, marker="s", facecolors="none",
                   edgecolors="cyan", linewidths=3)

    hallazgo = "masa" if caso["tipo"] == "mass" else "calcificaciones"
    ax.set_title(f"Caso {caso_numero} · {hallazgo}")
    ax.axis("off")
    plt.show()
    print(f"Caso: {caso['id']} · mama: {caso['lado']} · vista: {caso['vista']} · imagen: {alto} × {ancho} píxeles")
```

## Galería final · ocho casos ya marcados

Aquí están los tres casos que acabas de explorar y cinco ejemplos adicionales. Cada recuadro rojo rodea la región anotada en CBIS-DDSM. La etiqueta de benigno o maligno es el resultado registrado; el tamaño o color del recuadro no indica gravedad.

Ejecuta la celda: el cuaderno abrirá los ocho PNG y dibujará los recuadros usando las coordenadas de `casos.json`. La lámina se crea en ese momento; no es una imagen ya marcada.


```python
figura, ejes = plt.subplots(4, 2, figsize=(12, 22))

for numero, (caso, eje) in enumerate(zip(casos, ejes.flat), start=1):
    with Image.open(carpeta / "data/taller" / caso["imagen"]) as archivo:
        ancho_original, alto_original = archivo.size
        imagen = archivo.convert("L")
    imagen.thumbnail((500, 800))  # Reduce solo el tamaño de visualización.
    eje.imshow(imagen, cmap="gray", vmin=0, vmax=255)

    izquierda, arriba, derecha, abajo = caso["recuadro_referencia"]
    escala_x = imagen.width / ancho_original
    escala_y = imagen.height / alto_original
    eje.add_patch(Rectangle((izquierda * escala_x, arriba * escala_y),
                            (derecha - izquierda) * escala_x,
                            (abajo - arriba) * escala_y,
                            fill=False, edgecolor="red", linewidth=2.5))

    resultado = "maligno" if caso["patologia"] == "MALIGNANT" else "benigno"
    eje.set_title(f"{numero}. {caso['id']} · {resultado}")
    eje.axis("off")

plt.tight_layout()
plt.show()
```
