## 1. ¿Qué es una mama?

La mama tiene **grasa**, **tejido glandular** y **tejido fibroso** que le da soporte. En el tejido glandular están los lobulillos. Los conductos conectan esa zona con el pezón. Detrás de la mama está la pared torácica.

El esquema muestra dónde se encuentran estas partes. Es una guía para ubicarlas, no un dibujo a escala.


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Ellipse, Rectangle

figura, eje = plt.subplots(figsize=(9, 5))
eje.set_xlim(0, 10)
eje.set_ylim(0, 6)
eje.set_aspect("equal")
eje.axis("off")

# La pared torácica queda detrás de la mama.
eje.add_patch(Rectangle((1.0, 0.8), 0.4, 4.8, color="#75869b"))

# El óvalo claro representa el contorno de la mama.
eje.add_patch(Ellipse((5.0, 3.0), 7.0, 4.6,
                      facecolor="#f6d6bf", edgecolor="#9a6f65", linewidth=2))
eje.add_patch(Circle((8.5, 3.0), 0.16, color="#aa6b68"))

# Los óvalos rosados representan tejido glandular.
zonas_glandulares = [(3.2, 3.9), (4.1, 4.6), (4.3, 2.9),
                     (5.2, 3.8), (5.7, 2.3), (6.4, 3.4)]
for x, y in zonas_glandulares:
    eje.add_patch(Ellipse((x, y), 0.65, 0.9,
                          facecolor="#df8f9a", edgecolor="#ad6978"))
    eje.plot([x + 0.25, 8.35], [y, 3.0], color="#ad6978", linewidth=1.5)

eje.text(0.1, 5.75, "Pared torácica", color="#344054")
eje.text(2.1, 1.05, "Grasa", color="#704f44")
eje.text(3.0, 5.45, "Tejido glandular", color="#923a55")
eje.text(6.0, 4.85, "Conductos hacia el pezón", color="#923a55")
eje.set_title("Partes principales de la mama")
plt.show()
```

En el dibujo, la **grasa** ocupa buena parte del espacio claro, las zonas rosadas representan **tejido glandular** y las líneas llegan al **pezón**. En una mama real la distribución es más compleja y varía entre personas. [Anatomía de la mama, NCI](https://www.cancer.gov/types/breast/what-is-breast-cancer).


## 2. ¿Qué muestra una mamografía?

Una mamografía es una imagen de rayos X de la mama comprimida. En términos generales, la **grasa se ve más oscura** y el **tejido fibroglandular se ve más claro**. También pueden verse claras otras estructuras. Por eso, **una zona blanca no significa automáticamente cáncer**.

La imagen muestra una proyección de la mama: distintos tejidos pueden quedar superpuestos. Para entender un hallazgo, los profesionales comparan vistas y, cuando existen, estudios anteriores. En el taller verás la sigla **CC**, que indica una vista tomada desde arriba; **MLO** indica una vista oblicua. [Mamografías, NCI](https://www.cancer.gov/types/breast/screening/mammograms).


## 3. ¿Qué es una anomalía?

En este taller, *anomalía* significa **una región que merece una mirada más cuidadosa**. No es sinónimo de cáncer. En los tres casos posteriores aparecerán dos tipos de hallazgo:

| Hallazgo | Cómo lo describiremos al empezar |
|---|---|
| **Masa** | Una zona de aspecto compacto. Nos fijaremos en su forma y sus bordes. |
| **Calcificaciones** | Puntos claros pequeños. Nos fijaremos en si aparecen agrupados. |

El dibujo siguiente exagera estas formas para que puedas reconocer las palabras. **No reproduce la dificultad de encontrarlas en una mamografía real.**


```python
figura, ejes = plt.subplots(1, 2, figsize=(8, 4))

for eje in ejes:
    eje.set_xlim(0, 10)
    eje.set_ylim(0, 10)
    eje.set_aspect("equal")
    eje.axis("off")
    eje.add_patch(Ellipse((5, 5), 8, 9, facecolor="#454545"))

# Izquierda: una zona compacta.
ejes[0].add_patch(Ellipse((6.2, 5.8), 1.9, 1.6,
                          facecolor="#c8c8c8", edgecolor="#ededed"))
ejes[0].set_title("Masa · esquema")

# Derecha: varios puntos claros cercanos.
puntos = [(3.7, 5.4), (4.1, 5.9), (4.4, 5.2),
          (4.8, 5.7), (4.3, 6.3), (5.0, 6.1)]
for x, y in puntos:
    ejes[1].add_patch(Circle((x, y), 0.09, color="white"))
ejes[1].set_title("Calcificaciones · esquema")

plt.tight_layout()
plt.show()
```

A la izquierda hay **una zona compacta**; a la derecha, **varios puntos agrupados**. Esta diferencia ayuda a describir lo que vemos. Ni la forma de este dibujo ni el brillo de los puntos permiten decidir si un caso real es benigno o maligno. [Hallazgos en mamografías, NCI](https://www.cancer.gov/types/breast/screening/mammograms).
